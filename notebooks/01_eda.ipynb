# EDA — Financial Operations Analytics

Exploratory analysis of the synthetic transactions dataset.
Run after `make run` so processed data exists.

```python
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path

ROOT = Path("..")
df = pd.read_parquet(ROOT / "data/processed/transactions.parquet")
print(df.shape)
df.head()
```

```python
# Distributions
df["amount"].hist(bins=50)
plt.title("Amount distribution")
plt.xlabel("EUR")
plt.show()
```

```python
# Seasonality
monthly = df.groupby("year_month")["amount"].sum()
monthly.plot(kind="bar", figsize=(12, 4))
plt.title("Monthly net amount")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Category correlation (expense volume)
exp = df[df.amount < 0].pivot_table(index="year_month", columns="category", values="amount", aggfunc="sum").fillna(0)
print(exp.corr().round(2))
```
